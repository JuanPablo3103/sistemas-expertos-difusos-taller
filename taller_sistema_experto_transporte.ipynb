Sistema experto - como llegar a la universidad

Sistema hecho con CLIPSPY. Tiene 5 reglas y recomienda como moverse hacia la universidad segun el clima y el tiempo disponible.

```python
!pip install clipspy ipywidgets -q
import clips
import ipywidgets as widgets
from IPython.display import display, clear_output
```

Reglas:

1. si hay lluvia, necesito paraguas
2. si voy con afan, necesito salir rapido
3. si necesito salir rapido y ademas esta lloviendo, mejor tomar taxi
4. si tengo tiempo libre y el clima esta bueno, puedo ir caminando
5. si hace frio, necesito llevar chaqueta

```python
def crearSistema():
  sistemaExperto = clips.Environment()
  sistemaExperto.clear()

  reglaLluvia = "(defrule reglaLluvia (lluvia) => (assert(llevar_paraguas)))"
  reglaAfan = "(defrule reglaAfan (afan) => (assert(salir_rapido)))"
  reglaTaxi = "(defrule reglaTaxi (salir_rapido)(lluvia) => (assert(tomar_taxi)))"
  reglaCaminar = "(defrule reglaCaminar (tiempo_libre)(buen_clima) => (assert(ir_caminando)))"
  reglaFrio = "(defrule reglaFrio (frio) => (assert(llevar_chaqueta)))"

  sistemaExperto.build(reglaLluvia)
  sistemaExperto.build(reglaAfan)
  sistemaExperto.build(reglaTaxi)
  sistemaExperto.build(reglaCaminar)
  sistemaExperto.build(reglaFrio)

  return sistemaExperto

sistemaExperto = crearSistema()

#ver reglas cargadas
for r in sistemaExperto.rules():
  print(r)
```

Interfaz grafica

Se marca lo que aplica hoy y se le da click a Consultar.

```python
chkLluvia = widgets.Checkbox(value=False, description='Esta lloviendo')
chkAfan = widgets.Checkbox(value=False, description='Voy con afan')
chkTiempoLibre = widgets.Checkbox(value=False, description='Tengo tiempo libre')
chkBuenClima = widgets.Checkbox(value=False, description='El clima esta bueno')
chkFrio = widgets.Checkbox(value=False, description='Hace frio')

botonConsultar = widgets.Button(description='Consultar', button_style='success')
salida = widgets.Output()

def alConsultar(b):
  with salida:
    clear_output()
    sistema = crearSistema()

    if chkLluvia.value:
      sistema.assert_string("(lluvia)")
    if chkAfan.value:
      sistema.assert_string("(afan)")
    if chkTiempoLibre.value:
      sistema.assert_string("(tiempo_libre)")
    if chkBuenClima.value:
      sistema.assert_string("(buen_clima)")
    if chkFrio.value:
      sistema.assert_string("(frio)")

    sistema.run()

    huboRecomendacion = False
    for fact in sistema.facts():
      f = str(fact)
      if "tomar_taxi" in f:
        print("Recomendacion: toma un taxi, esta lloviendo y no hay tiempo")
        huboRecomendacion = True
      if "ir_caminando" in f:
        print("Recomendacion: puedes ir caminando, hay tiempo y el clima esta bueno")
        huboRecomendacion = True
      if "llevar_paraguas" in f:
        print("Recomendacion: lleva paraguas por si acaso")
        huboRecomendacion = True
      if "llevar_chaqueta" in f:
        print("Recomendacion: lleva chaqueta, hace frio")
        huboRecomendacion = True

    if not huboRecomendacion:
      print("Con estas condiciones el sistema no tiene una recomendacion clara todavia")

botonConsultar.on_click(alConsultar)

display(chkLluvia, chkAfan, chkTiempoLibre, chkBuenClima, chkFrio, botonConsultar, salida)
```
